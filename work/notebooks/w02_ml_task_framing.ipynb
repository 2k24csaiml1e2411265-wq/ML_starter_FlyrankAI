## 5. Why ML beats a fixed rule here

A fixed rule, such as refreshing pages every six months or using a single traffic threshold, cannot capture the complex relationships between search performance, user behavior, and content quality.

Machine learning can learn patterns from multiple features simultaneously and identify pages that are at higher risk of declining performance.

The model provides decision-support by prioritizing pages for review instead of relying on simple manual rules.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Example target sketch

if "clicks" in df.columns:
    threshold = df["clicks"].median()
    df["target_decline"] = (df["clicks"] < threshold).astype(int)
else:
    # Placeholder example if no clicks column exists
    df["target_decline"] = 0

print(df[["target_decline"]].head())
```
